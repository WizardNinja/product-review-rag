```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Reviews.csv")
```

```python
print(f"Dataset shape: {df.shape}")
print(f"\nColumns: {df.columns.tolist()}")
print(f"\nData types:\n{df.dtypes}")
```

```python
subset_size = 2000
df_subset = df.groupby('Score').apply(
    lambda x: x.sample(min(len(x), subset_size // 5), random_state=42)
).reset_index(drop=True)

print(f"Subset size: {len(df_subset)}")
print(f"Rating distribution in subset:\n{df_subset['Score'].value_counts().sort_index()}")

df_subset.to_csv('reviews_subset.csv', index=False)
print("\nSaved to reviews_subset.csv")
```
